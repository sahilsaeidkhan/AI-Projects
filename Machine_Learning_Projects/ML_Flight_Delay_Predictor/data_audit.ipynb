```python
def add_date_features(data):
    data = data.copy()

    data["DAY_OF_MONTH"] = data["FL_DATE"].dt.day
    data["DAY_OF_WEEK"] = data["FL_DATE"].dt.dayofweek
    data["WEEK_OF_MONTH"] = ((data["FL_DATE"].dt.day - 1) // 7) + 1

    return data
```

```python
train_df = add_date_features(train_df)
val_df = add_date_features(val_df)
test_df = add_date_features(test_df)
```

```python
def add_time_features(data):
    data = data.copy()

    # Convert HHMM format into hour and minute
    data["DEP_HOUR"] = data["CRS_DEP_TIME"] // 100
    data["DEP_MINUTE"] = data["CRS_DEP_TIME"] % 100

    # Same for scheduled arrival
    data["ARR_HOUR"] = data["CRS_ARR_TIME"] // 100
    data["ARR_MINUTE"] = data["CRS_ARR_TIME"] % 100

    return data
```

```python
train_df = add_time_features(train_df)
val_df = add_time_features(val_df)
test_df = add_time_features(test_df)
```
