```python
# ============================================================
# LEAKAGE-SAFE HISTORICAL FEATURE ENGINEERING
# ============================================================

# Started again from the original clean dataset
df_clean = df[
    (df["CANCELLED"] == 0) &
    (df["DIVERTED"] == 0)
].copy()

# Convert date
df_clean["FL_DATE"] = pd.to_datetime(df_clean["FL_DATE"])

# Sort chronologically
df_clean = df_clean.sort_values(
    ["FL_DATE", "CRS_DEP_TIME", "OP_CARRIER_FL_NUM"]
).reset_index(drop=True)


# ------------------------------------------------------------
# 1. Basic features
# ------------------------------------------------------------

df_clean["DAY_OF_MONTH"] = df_clean["FL_DATE"].dt.day
df_clean["DAY_OF_WEEK"] = df_clean["FL_DATE"].dt.dayofweek
df_clean["WEEK_OF_MONTH"] = (
    (df_clean["FL_DATE"].dt.day - 1) // 7
) + 1

df_clean["DEP_HOUR"] = df_clean["CRS_DEP_TIME"] // 100
df_clean["DEP_MINUTE"] = df_clean["CRS_DEP_TIME"] % 100

df_clean["ARR_HOUR"] = df_clean["CRS_ARR_TIME"] // 100
df_clean["ARR_MINUTE"] = df_clean["CRS_ARR_TIME"] % 100


# ------------------------------------------------------------
# 2. Create route
# ------------------------------------------------------------

df_clean["ROUTE"] = (
    df_clean["ORIGIN_CITY_NAME"]
    + " -> "
    + df_clean["DEST_CITY_NAME"]
)


# ------------------------------------------------------------
# 3. Function to create leakage-safe historical delay rate
# ------------------------------------------------------------

def add_historical_delay_rate(data, group_columns, feature_name):
    
    # Daily statistics
    daily = (
        data
        .groupby(group_columns + ["FL_DATE"])["ARR_DEL15"]
        .agg(["sum", "count"])
        .reset_index()
    )

    # Sort by group + date
    daily = daily.sort_values(
        group_columns + ["FL_DATE"]
    ).copy()

    # Cumulative values
    daily["cumulative_delays"] = (
        daily
        .groupby(group_columns)["sum"]
        .cumsum()
    )

    daily["cumulative_flights"] = (
        daily
        .groupby(group_columns)["count"]
        .cumsum()
    )

    # Remove CURRENT day's information
    daily["previous_delays"] = (
        daily["cumulative_delays"]
        - daily["sum"]
    )

    daily["previous_flights"] = (
        daily["cumulative_flights"]
        - daily["count"]
    )

    # Historical rate using ONLY previous days
    daily[feature_name] = (
        daily["previous_delays"]
        / daily["previous_flights"]
    )

    # Keep only columns needed for merging
    merge_columns = group_columns + [
        "FL_DATE",
        feature_name
    ]

    daily = daily[merge_columns]

    # Merge back into original data
    data = data.merge(
        daily,
        on=group_columns + ["FL_DATE"],
        how="left"
    )

    return data


# ------------------------------------------------------------
# 4. Airline historical delay rate
# ------------------------------------------------------------

df_clean = add_historical_delay_rate(
    df_clean,
    ["OP_UNIQUE_CARRIER"],
    "carrier_historical_delay_rate"
)


# ------------------------------------------------------------
# 5. Origin historical delay rate
# ------------------------------------------------------------

df_clean = add_historical_delay_rate(
    df_clean,
    ["ORIGIN_CITY_NAME"],
    "origin_historical_delay_rate"
)


# ------------------------------------------------------------
# 6. Destination historical delay rate
# ------------------------------------------------------------

df_clean = add_historical_delay_rate(
    df_clean,
    ["DEST_CITY_NAME"],
    "destination_historical_delay_rate"
)


# ------------------------------------------------------------
# 7. Route historical delay rate
# ------------------------------------------------------------

df_clean = add_historical_delay_rate(
    df_clean,
    ["ROUTE"],
    "route_historical_delay_rate"
)


# ------------------------------------------------------------
# 8. Departure-hour historical delay rate
# ------------------------------------------------------------

df_clean = add_historical_delay_rate(
    df_clean,
    ["DEP_HOUR"],
    "departure_hour_historical_delay_rate"
)


# ------------------------------------------------------------
# 9. Create chronological train / validation / test
# ------------------------------------------------------------

train_df = df_clean[
    df_clean["FL_DATE"] <= "2025-01-21"
].copy()

val_df = df_clean[
    (df_clean["FL_DATE"] >= "2025-01-22") &
    (df_clean["FL_DATE"] <= "2025-01-26")
].copy()

test_df = df_clean[
    df_clean["FL_DATE"] >= "2025-01-27"
].copy()


# ------------------------------------------------------------
# 10. Fill missing historical values
# ------------------------------------------------------------
# Missing values occur mainly when there is no previous history.
# Use ONLY the training-set overall delay rate as fallback.

overall_delay_rate = train_df["ARR_DEL15"].mean()

historical_features = [
    "carrier_historical_delay_rate",
    "origin_historical_delay_rate",
    "destination_historical_delay_rate",
    "route_historical_delay_rate",
    "departure_hour_historical_delay_rate"
]

for feature in historical_features:
    train_df[feature] = train_df[feature].fillna(
        overall_delay_rate
    )

    val_df[feature] = val_df[feature].fillna(
        overall_delay_rate
    )

    test_df[feature] = test_df[feature].fillna(
        overall_delay_rate
    )


# ------------------------------------------------------------
# 11. Define final features
# ------------------------------------------------------------

feature_columns = [
    # Date
    "DAY_OF_MONTH",
    "DAY_OF_WEEK",
    "WEEK_OF_MONTH",

    # Flight information
    "OP_UNIQUE_CARRIER",
    "ORIGIN_CITY_NAME",
    "ORIGIN_STATE_ABR",
    "DEST_CITY_NAME",
    "DEST_STATE_ABR",

    # Time
    "DEP_HOUR",
    "DEP_MINUTE",
    "ARR_HOUR",
    "ARR_MINUTE",

    # Numeric
    "CRS_ELAPSED_TIME",
    "DISTANCE",

    # Historical features
    "carrier_historical_delay_rate",
    "origin_historical_delay_rate",
    "destination_historical_delay_rate",
    "route_historical_delay_rate",
    "departure_hour_historical_delay_rate"
]

target = "ARR_DEL15"


# ------------------------------------------------------------
# 12. Create X and y
# ------------------------------------------------------------

X_train = train_df[feature_columns]
y_train = train_df[target]

X_val = val_df[feature_columns]
y_val = val_df[target]

X_test = test_df[feature_columns]
y_test = test_df[target]


# ------------------------------------------------------------
# 13. Verify everything
# ------------------------------------------------------------

print("Train shape:", X_train.shape)
print("Validation shape:", X_val.shape)
print("Test shape:", X_test.shape)

print("\nHistorical feature missing values:")
print(
    X_train[historical_features]
    .isna()
    .sum()
)

print("\nHistorical feature statistics:")
print(
    X_train[historical_features]
    .describe()
)
```

```python
# ============================================================
# IMPROVED FEATURE ENGINEERING
# ============================================================

# ------------------------------------------------------------
# 1. Create route features
# ------------------------------------------------------------

for data in [train_df, val_df, test_df]:

    data["ROUTE"] = (
        data["ORIGIN_CITY_NAME"]
        + " -> "
        + data["DEST_CITY_NAME"]
    )

    data["AIRLINE_ROUTE"] = (
        data["OP_UNIQUE_CARRIER"]
        + "_"
        + data["ROUTE"]
    )


# ------------------------------------------------------------
# 2. Cyclic departure-time features
# ------------------------------------------------------------

for data in [train_df, val_df, test_df]:

    # Convert departure time to minutes after midnight
    data["DEP_MINUTES"] = (
        data["DEP_HOUR"] * 60
        + data["DEP_MINUTE"]
    )

    # 1440 minutes in a day
    data["DEP_TIME_SIN"] = np.sin(
        2 * np.pi * data["DEP_MINUTES"] / 1440
    )

    data["DEP_TIME_COS"] = np.cos(
        2 * np.pi * data["DEP_MINUTES"] / 1440
    )


# ------------------------------------------------------------
# 3. Cyclic day-of-week features
# ------------------------------------------------------------

for data in [train_df, val_df, test_df]:

    data["DOW_SIN"] = np.sin(
        2 * np.pi * data["DAY_OF_WEEK"] / 7
    )

    data["DOW_COS"] = np.cos(
        2 * np.pi * data["DAY_OF_WEEK"] / 7
    )


# ------------------------------------------------------------
# 4. Define improved feature set
# ------------------------------------------------------------

feature_columns_v2 = [

    # Date
    "DAY_OF_MONTH",
    "DAY_OF_WEEK",
    "WEEK_OF_MONTH",

    # Airline
    "OP_UNIQUE_CARRIER",

    # Location
    "ORIGIN_CITY_NAME",
    "ORIGIN_STATE_ABR",
    "DEST_CITY_NAME",
    "DEST_STATE_ABR",

    # Route
    "ROUTE",
    "AIRLINE_ROUTE",

    # Time
    "DEP_HOUR",
    "DEP_MINUTE",
    "ARR_HOUR",
    "ARR_MINUTE",

    "DEP_TIME_SIN",
    "DEP_TIME_COS",

    "DOW_SIN",
    "DOW_COS",

    # Flight characteristics
    "CRS_ELAPSED_TIME",
    "DISTANCE",

    # Historical information
    "carrier_historical_delay_rate",
    "origin_historical_delay_rate",
    "destination_historical_delay_rate",
    "route_historical_delay_rate",
    "departure_hour_historical_delay_rate"
]


# ------------------------------------------------------------
# 5. Create X / y
# ------------------------------------------------------------

X_train_v2 = train_df[feature_columns_v2]
y_train_v2 = train_df["ARR_DEL15"]

X_val_v2 = val_df[feature_columns_v2]
y_val_v2 = val_df["ARR_DEL15"]

X_test_v2 = test_df[feature_columns_v2]
y_test_v2 = test_df["ARR_DEL15"]


# ------------------------------------------------------------
# 6. Define preprocessing
# ------------------------------------------------------------

numeric_features_v2 = [
    "DAY_OF_MONTH",
    "DAY_OF_WEEK",
    "WEEK_OF_MONTH",

    "DEP_HOUR",
    "DEP_MINUTE",
    "ARR_HOUR",
    "ARR_MINUTE",

    "DEP_TIME_SIN",
    "DEP_TIME_COS",

    "DOW_SIN",
    "DOW_COS",

    "CRS_ELAPSED_TIME",
    "DISTANCE",

    "carrier_historical_delay_rate",
    "origin_historical_delay_rate",
    "destination_historical_delay_rate",
    "route_historical_delay_rate",
    "departure_hour_historical_delay_rate"
]

categorical_features_v2 = [
    "OP_UNIQUE_CARRIER",
    "ORIGIN_CITY_NAME",
    "ORIGIN_STATE_ABR",
    "DEST_CITY_NAME",
    "DEST_STATE_ABR",
    "ROUTE",
    "AIRLINE_ROUTE"
]


preprocessor_v2 = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features_v2
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features_v2
        )
    ]
)


# ------------------------------------------------------------
# 7. Check shapes
# ------------------------------------------------------------

print("Train:", X_train_v2.shape)
print("Validation:", X_val_v2.shape)
print("Test:", X_test_v2.shape)

print("\nFeatures:", len(feature_columns_v2))
```

```python
xgb_model_v2 = Pipeline(
    steps=[
        ("preprocessor", preprocessor_v2),

        ("model", XGBClassifier(
            n_estimators=250,
            max_depth=6,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42,
            n_jobs=-1,
            tree_method="hist"
        ))
    ]
)
```

```python
# ============================================================
# CLASS-WEIGHTED XGBOOST
# ============================================================

from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline


# ------------------------------------------------------------
# 1. Build weighted XGBoost
# ------------------------------------------------------------

weighted_xgb_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor_v2),

        ("model", XGBClassifier(
            n_estimators=200,
            max_depth=8,
            learning_rate=0.05,
            min_child_weight=1,
            subsample=0.7,
            colsample_bytree=0.8,

            # Important: give more importance to delays
            scale_pos_weight=scale_pos_weight,

            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42,
            n_jobs=-1,
            tree_method="hist"
        ))
    ]
)


# ------------------------------------------------------------
# 2. Train
# ------------------------------------------------------------

weighted_xgb_model.fit(
    X_train_v2,
    y_train_v2
)


# ------------------------------------------------------------
# 3. Validation predictions
# ------------------------------------------------------------

weighted_probabilities = (
    weighted_xgb_model
    .predict_proba(X_val_v2)[:, 1]
)

# Default threshold
weighted_predictions = (
    weighted_probabilities >= 0.50
).astype(int)


# ------------------------------------------------------------
# 4. Evaluate at threshold 0.50
# ------------------------------------------------------------

print("CLASS-WEIGHTED XGBOOST")
print("=" * 45)

print(
    "Accuracy :",
    accuracy_score(
        y_val_v2,
        weighted_predictions
    )
)

print(
    "Precision:",
    precision_score(
        y_val_v2,
        weighted_predictions,
        zero_division=0
    )
)

print(
    "Recall   :",
    recall_score(
        y_val_v2,
        weighted_predictions,
        zero_division=0
    )
)

print(
    "F1 Score :",
    f1_score(
        y_val_v2,
        weighted_predictions,
        zero_division=0
    )
)

print(
    "ROC-AUC  :",
    roc_auc_score(
        y_val_v2,
        weighted_probabilities
    )
)

print(
    "PR-AUC   :",
    average_precision_score(
        y_val_v2,
        weighted_probabilities
    )
)


# ------------------------------------------------------------
# 5. Find best threshold on VALIDATION
# ------------------------------------------------------------

weighted_threshold_results = []

for threshold in np.arange(0.05, 0.56, 0.01):

    predictions = (
        weighted_probabilities >= threshold
    ).astype(int)

    weighted_threshold_results.append({

        "Threshold": round(threshold, 2),

        "Accuracy": accuracy_score(
            y_val_v2,
            predictions
        ),

        "Precision": precision_score(
            y_val_v2,
            predictions,
            zero_division=0
        ),

        "Recall": recall_score(
            y_val_v2,
            predictions,
            zero_division=0
        ),

        "F1": f1_score(
            y_val_v2,
            predictions,
            zero_division=0
        )
    })


weighted_threshold_df = pd.DataFrame(
    weighted_threshold_results
)


# ------------------------------------------------------------
# 6. Best F1 threshold
# ------------------------------------------------------------

best_weighted_threshold = weighted_threshold_df.loc[
    weighted_threshold_df["F1"].idxmax()
]


print("\nBEST THRESHOLD")
print("=" * 45)

print(best_weighted_threshold)


# ------------------------------------------------------------
# 7. Top threshold choices
# ------------------------------------------------------------

print("\nTOP 10 THRESHOLDS")
print("=" * 45)

print(
    weighted_threshold_df
    .sort_values("F1", ascending=False)
    .head(10)
)
```
