```python
df = df.dropna(subset=[
    "CustomerID",
    "Gender",
    "Location",
    "Tenure_Months",
    "Transaction_ID",
    "Transaction_Date",
    "Product_SKU",
    "Product_Description",
    "Quantity",
    "Avg_Price",
    "Delivery_Charges",
    "Coupon_Status",
    "GST",
    "Date",
    "Offline_Spend",
    "Online_Spend"
])
```

```python
df = df[
    ~(
        (df["Coupon_Status"] == "Used") &
        (df["Coupon_Code"].isnull())
    )
]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))

ax = sns.countplot(data=df, x="Gender")

for container in ax.containers:
    ax.bar_label(container)

plt.title("Number of Purchases by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Purchases")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="Month_Name",
    order=month_order
)

plt.xticks(rotation=45)
plt.title("Purchases by Month")
plt.xlabel("Month")
plt.ylabel("Number of Purchases")

plt.show()
```

```python
day_order = [
    "Monday","Tuesday","Wednesday",
    "Thursday","Friday","Saturday","Sunday"
]

plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="Day_Name",
    order=day_order
)

plt.title("Purchases by Day of Week")
plt.xlabel("Day")
plt.ylabel("Number of Purchases")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convert Date to datetime (skip if already done)
df["Date"] = pd.to_datetime(df["Date"])

# Extract month name
df["Month_Name"] = df["Date"].dt.month_name()

# Order months correctly
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x="Month_Name",
    hue="Gender",
    order=month_order
)

plt.title("Monthly Purchases by Gender", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=45)
plt.legend(title="Gender")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,6))

sns.countplot(
    data=df,
    x="Product_Category",
    hue="Gender",
    order=df["Product_Category"].value_counts().index
)

plt.title("Product Category Purchased by Gender")
plt.xlabel("Product Category")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=45)
plt.legend(title="Gender")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

quantity_df = (
    df.groupby(["Product_Category", "Gender"])["Quantity"]
      .sum()
      .reset_index()
)

plt.figure(figsize=(14,6))

sns.barplot(
    data=quantity_df,
    x="Product_Category",
    y="Quantity",
    hue="Gender"
)

plt.title("Total Quantity Purchased by Gender")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Purchased")
plt.xticks(rotation=45)
plt.legend(title="Gender")

plt.tight_layout()
plt.show()
```
